```python
import sqlite3
import pandas as pd

# Conexión en memoria (se pierde al cerrar el notebook)
conn = sqlite3.connect(":restaurante.db:")
cur = conn.cursor() #objeto para ejecutar comandos de SQL

def query(sql, params=()):
    """Ejecuta un SELECT y devuelve un DataFrame para verlo bonito."""
    return pd.read_sql_query(sql, conn, params=params)

def run(sql, params=()):
    """Ejecuta INSERT/UPDATE/DELETE/DDL y hace commit."""
    cur.execute(sql, params)
    conn.commit()
    print("OK ->", sql.strip().splitlines()[0], "...")
```

Ejercicio 1

Una cadena de restaurantes necesita registrar en su base de datos el menú de platillos que ofrece. Crea
una tabla llamada comida que cumpla con las siguientes especificaciones:

```python
cur.executescript('''
DROP TABLE IF EXISTS comida;

CREATE TABLE comida (
    id_platillo INTEGER PRIMARY KEY,
    nombre VARCHAR(50) NOT NULL,
    categoria VARCHAR(50),
    precio DECIMAL(6,2) NOT NULL,
    disponible BOOLEAN DEFAULT TRUE
);
''')
conn.commit()
print("Tabla recreada con columna 'disponible' correcta")
```

```python
# INSERT INTO

run('''
INSERT INTO comida (id_platillo, nombre, categoria, precio, disponible) VALUES
    (1, 'Tacos al pastor', 'plato fuerte', 85.00, 1),
    (2, 'Ensalada César', 'entrada', 65.50, 1),
    (3, 'Flan napolitano', 'postre', 45.00, 1);
''')

query("SELECT * FROM comida")
```

```python
# SELECT

query("SELECT * FROM comida")
```

```python
# UPDATE

run("UPDATE comida SET precio = precio * 1.10 WHERE categoria = 'plato fuerte';")

query("SELECT id_platillo, nombre, precio FROM comida WHERE categoria = 'plato fuerte'")
```

```python
# DELETE

run("DELETE FROM comida WHERE categoria = 'postre';")

query("SELECT * FROM comida")
```

Ejemplo 2 — Escuela (tabla alumnos)

Una escuela quiere registrar la información básica de sus alumnos.

```python
cur.executescript('''
DROP TABLE IF EXISTS alumnos;

CREATE TABLE alumnos (
    id_alumno INTEGER PRIMARY KEY,
    nombre VARCHAR(50) NOT NULL,
    grado VARCHAR(10),
    edad INT NOT NULL,
    promedio DECIMAL(4,2) DEFAULT 0.0
);
''')
conn.commit()
print("Tabla 'alumnos' creada")
```

```python
# INSERT INTO
run('''
INSERT INTO alumnos (nombre, grado, edad, promedio) VALUES
    ('Ana López', '3ro A', 9, 8.5),
    ('Carlos Ruiz', '5to B', 11, 7.8),
    ('María Sánchez', '3ro A', 8, 9.2);
''')

query("SELECT * FROM alumnos")
```

```python
# UPDATE
run("UPDATE alumnos SET promedio = promedio + 0.3 WHERE grado = '3ro A';")

query("SELECT id_alumno, nombre, grado, promedio FROM alumnos WHERE grado = '3ro A'")
```

```python
# DELETE
run("DELETE FROM alumnos WHERE nombre = 'Carlos Ruiz';")

query("SELECT * FROM alumnos")
```

Ejemplo 3 — Tiendita escolar

La tiendita de la escuela necesita llevar el control de los productos que vende.



```python
cur.executescript('''
DROP TABLE IF EXISTS productos;

CREATE TABLE productos (
    id_producto INTEGER PRIMARY KEY,
    nombre VARCHAR(50) NOT NULL,
    tipo VARCHAR(30),
    precio DECIMAL(6,2) NOT NULL,
    existencia INT DEFAULT 0
);
''')
conn.commit()
print("Tabla 'productos' creada")
```

```python
# INSERT INTO
run('''
INSERT INTO productos (nombre, tipo, precio, existencia) VALUES
    ('Papas fritas', 'snack', 15.00, 30),
    ('Jugo de manzana', 'bebida', 12.50, 25),
    ('Chocolate', 'dulce', 10.00, 40);
''')

query("SELECT * FROM productos")
```

```python
# UPDATE
run("UPDATE productos SET existencia = existencia - 10 WHERE nombre = 'Chocolate';")

query("SELECT id_producto, nombre, existencia FROM productos WHERE nombre = 'Chocolate'")
```

```python
# DELETE
run("DELETE FROM productos WHERE nombre = 'Jugo de manzana';")

query("SELECT * FROM productos")
```
